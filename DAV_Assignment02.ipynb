```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("/content/earthquake_1995-2023.csv")

# Create histogram
plt.figure(figsize=(10, 6))

plt.hist(
    df["magnitude"].dropna(),
    bins=20,
    color="pink",
    edgecolor="black",
    alpha=0.8
)

# Title and labels
plt.title(
    "Distribution of Earthquake Magnitudes",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Earthquake Magnitude", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

# Grid
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Load the dataset
df = pd.read_csv("earthquake_1995-2023.csv")

# Count earthquakes for each country
country_counts = df["country"].value_counts().head(10)

# Create colour mapping
norm = mpl.colors.Normalize(
    vmin=country_counts.min(),
    vmax=country_counts.max()
)

# Change the colormap here
cmap = mpl.cm.plasma

bar_colors = cmap(norm(country_counts.values))

# Plot bar graph
plt.figure(figsize=(10, 6))

plt.bar(
    country_counts.index,
    country_counts.values,
    color=bar_colors
)

# Add colour bar
sm = mpl.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])

plt.colorbar(
    sm,
    ax=plt.gca(),
    label="Number of Earthquakes"
)

# Labels and title
plt.title("Top 10 Countries by Number of Earthquakes")
plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")
plt.xticks(rotation=45)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Load the dataset
df = pd.read_csv("earthquake_1995-2023.csv")

# Count earthquakes for each country
country_counts = df["country"].value_counts().head(5)

# Create different shades of pink
cmap = mpl.cm.RdPu
colors = cmap([0.3, 0.45, 0.6, 0.75, 0.9])

# Create figure
plt.figure(figsize=(6, 6))

# Create donut chart
plt.pie(
    country_counts.values,
    labels=country_counts.index,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.5}
)

# Add title
plt.title("Top 5 Countries by Number of Earthquakes")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("earthquake_1995-2023.csv")

# Convert date_time to datetime
df["date_time"] = pd.to_datetime(df["date_time"])

# Extract year
df["year"] = df["date_time"].dt.year

# Count earthquakes for each year
yearly_count = df["year"].value_counts().sort_index()

# Line plot
plt.figure(figsize=(10, 6))

plt.plot(
    yearly_count.index,
    yearly_count.values,
    marker="o"
)

plt.title("Number of Earthquakes by Year")
plt.xlabel("Year")
plt.ylabel("Number of Earthquakes")

plt.grid(True, alpha=0.3)
plt.show()
```
